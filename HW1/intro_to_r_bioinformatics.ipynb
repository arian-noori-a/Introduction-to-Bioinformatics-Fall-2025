# **🧩 PART 8 — Writing Functions in R**

Define reusable functions, work with arguments, return values, and simple checks.

8.1 Function Basics

```r
# Example: a simple function to compute GC content of a DNA string
gc_content <- function(dna) {
  # Convert to upper just in case
  s <- toupper(dna)
  n <- nchar(s)
  if (n == 0) return(NA_real_)
  gc <- sum(unlist(strsplit(s, "")) %in% c("G", "C"))
  return(100 * gc / n)
}

# Try it
gc_content("ATGCGT")
gc_content("")
```

Write a function nucleotide_counts(seq) that returns a named vector with counts of A, T, C, G.

```r
nucleotide_counts <- function(seq) {
  s <- toupper(seq)
  chars <- unlist(strsplit(s, ""))

  c(
    A = sum(chars == "A"),
    T = sum(chars == "T"),
    C = sum(chars == "C"),
    G = sum(chars == "G")
  )
}

seq1 <- "ACGTacgtaAATN"; nucleotide_counts(seq1)
seqs <- c(s1="ATGCA", s2="ttttggccaa", s3="NNNNacgtACGT")
counts_mat <- t(sapply(seqs, nucleotide_counts)); counts_mat

res <- as.data.frame(counts_mat); res$total <- rowSums(res); res$GC <- (res$G + res$C) / res$total; res
```

8.2 Arguments with Defaults & Input Validation

```r
# Example: k-mer counter with default k = 3
kmer_count <- function(seq, k = 3) {
  s <- toupper(seq)
  n <- nchar(s)
  if (!is.numeric(k) || length(k) != 1 || k < 1) {
    stop("k must be a positive integer length 1.")
  }
  if (k > n) return(integer(0))
  kmers <- sapply(1:(n - k + 1), function(i) substr(s, i, i + k - 1))
  tab <- table(kmers)
  return(tab[order(names(tab))])
}

# Try it
kmer_count("ATGCGT", k = 2)
```

Write a function safe_mean(x, na_rm = TRUE) that:

* checks x is numeric,

* if na_rm is TRUE, ignores NAs,

* returns NA if all values are NA or x is length 0.

```r
safe_mean <- function(x, na_rm = TRUE) {
  if (!is.numeric(x)) stop("x must be numeric")
  if (length(x) == 0) return(NA_real_)
  m <- mean(x, na.rm = na_rm)
  if (is.nan(m)) return(NA_real_)
  m
}

library(testthat)

test_that("computes mean of numeric vector", {
  expect_equal(safe_mean(c(1, 2, 3)), 2)
})
```

8.3 Returning Multiple Values (Lists) & Simple Tests

```r
# Example: summary of a numeric vector
vec_summary <- function(x) {
  if (!is.numeric(x)) stop("x must be numeric.")
  if (length(x) == 0) return(list(mean = NA_real_, sd = NA_real_, n = 0))
  out <- list(
    mean = mean(x, na.rm = TRUE),
    sd   = sd(x, na.rm = TRUE),
    n    = sum(!is.na(x))
  )
  return(out)
}

vec_summary(c(1, 2, 3, NA))
```

# **🎯 Mini-Tasks (Functions)**

Mini-Task F1 — Reusable QC function
Write a function qc_read(read_seq, min_len = 50, allowed = c("A","C","G","T","N")) that:

* checks length ≥ min_len,

* returns TRUE/FALSE,

* ensures all characters belong to allowed.

```r
qc_read <- function(read_seq, min_len = 50, allowed = c("A","C","G","T","N")) {
  s <- toupper(read_seq)
  if (nchar(s) < min_len) return(FALSE)
  chars <- unlist(strsplit(s, ""))
  all(chars %in% allowed)
}

# tests
qc_read("ACTGACTG", min_len = 5)
qc_read("ACTGXXXX", min_len = 5)
qc_read(paste(rep("A",60), collapse=""), min_len = 50)
```

# **Basic Statistics: t-test & Multiple Testing (BH/FDR)**

run simple tests and control FDR (like BH in RNA-seq workflows).

```r
set.seed(42)
grp <- rep(c("treated","control"), each = 10)
x   <- rnorm(20, mean = ifelse(grp=="treated", 1, 0), sd = 1)

# Two-sample t-test
t.test(x[grp=="treated"], x[grp=="control"])
```

```r
# START HERE
set.seed(42)

n_genes <- 100
n_per   <- 10
shift   <- c(rep(1, n_genes/2), rep(0, n_genes/2))  # +1 for DE, 0 otherwise

# Simulate matrix: rows = genes, first n_per cols = treated, next n_per = control
X <- matrix(rnorm(n_genes * 2 * n_per, mean = 0, sd = 1), nrow = n_genes)
X[, 1:n_per] <- X[, 1:n_per] + shift  # apply shift to treated for DE genes

# 1) Per-gene Welch t-test p-values
pvals <- sapply(seq_len(n_genes), function(g)
  t.test(X[g, 1:n_per], X[g, (n_per + 1):(2 * n_per)])$p.value
)

# 2) BH-adjusted p-values
padj <- p.adjust(pvals, method = "BH")

# 3) Count significant at FDR < 0.05
sig <- padj < 0.05
n_sig <- sum(sig)
tp <- sum(sig & shift == 1)
fp <- sum(sig & shift == 0)

cat(sprintf("Significant (FDR < 0.05): %d\n", n_sig))
cat(sprintf("  True positives:  %d\n", tp))
cat(sprintf("  False positives: %d\n", fp))

# Optional: results table
results <- data.frame(
  gene = paste0("gene_", seq_len(n_genes)),
  is_DE = shift == 1,
  pval = pvals,
  padj = padj,
  significant = sig
)
head(results, 10)
# END HERE
```

# **Correlation, Heatmap, and Clustering**

quick QC / similarity checks between samples.

```r
# Build a small expression matrix (genes x samples)
set.seed(1)
mat <- matrix(rnorm(200), nrow = 20)  # 20 genes, 10 samples
colnames(mat) <- paste0("S", 1:10)

# Correlation between samples
cors <- cor(mat)
# Heatmap (base R)
# heatmap(cors)
```

```r
# START HERE

# 1) Scale each gene (row) to mean 0 / sd 1
mat_scaled <- t(scale(t(mat), center = TRUE, scale = TRUE))

# 2) Compute correlation across samples on the scaled matrix
cors_scaled <- cor(mat_scaled)

# 3) Hierarchical clustering of samples and plot dendrogram
d_samples <- dist(t(mat_scaled))            # distances between samples (columns)
hc_samples <- hclust(d_samples, method = "complete")
plot(hc_samples, main = "Sample clustering (scaled expression)")

heatmap(cors_scaled,
        Rowv = as.dendrogram(hc_samples),
        Colv = as.dendrogram(hc_samples),
        symm = TRUE,
        main = "Sample correlation (scaled)")

# END HERE
```

# **PCA (Principal Components) with prcomp**

dimension reduction for expression matrices.

```r
set.seed(7)
X <- matrix(rnorm(300), nrow = 30)  # 30 genes x 10 samples
colnames(X) <- paste0("S", 1:10)

# PCA expects samples in rows (or set transpose accordingly)
pca <- prcomp(t(X), scale. = TRUE)
summary(pca)$importance[ , 1:3]  # variance explained
```

```r
# START HERE

# 1) Run PCA on t(X) with scaling
pca <- prcomp(t(X), scale. = TRUE)

# 2) Extract PC1 and PC2 scores and make a basic scatter
pc1 <- pca$x[, 1]
pc2 <- pca$x[, 2]

# Label axes with % variance explained
ve <- summary(pca)$importance[2, 1:2] * 100
plot(pc1, pc2,
     xlab = sprintf("PC1 (%.1f%%)", ve[1]),
     ylab = sprintf("PC2 (%.1f%%)", ve[2]),
     main = "PCA of samples",
     pch = 19)
text(pc1, pc2, labels = colnames(X), pos = 3, cex = 0.8)

# END HERE
```
